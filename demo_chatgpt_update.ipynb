## Demo chart added by ChatGPT

Run the next cell in Colab. It creates a small table, draws a bar chart, and saves the chart image inside the Colab runtime.

```python
#@title Demo chart: sample quarterly values
import os
from datetime import datetime, timezone

import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# Small demo dataset. Replace these numbers later with real data if needed.
data = pd.DataFrame({
    'Quarter': ['Q1', 'Q2', 'Q3', 'Q4'],
    'Value': [120, 150, 135, 180]
})

display(data)

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(data['Quarter'], data['Value'])
ax.set_title('Demo chart: sample quarterly values')
ax.set_xlabel('Quarter')
ax.set_ylabel('Value')
ax.grid(axis='y', alpha=0.3)

OUTDIR = '/content/demo_outputs'
os.makedirs(OUTDIR, exist_ok=True)
chart_path = os.path.join(OUTDIR, 'demo_quarterly_values_chart.png')
fig.savefig(chart_path, dpi=150, bbox_inches='tight')
plt.show()

print('Chart saved inside this Colab run at:', chart_path)
print('Chart created at UTC:', datetime.now(timezone.utc).strftime('%Y-%m-%d %H:%M:%S'))
```

## Second update added by ChatGPT

This section was pushed after you already had the Colab tab open. Use it to test whether reopening or refreshing the notebook pulls the latest GitHub version.

```python
#@title Second update: cumulative line chart
import os
from datetime import datetime, timezone

import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

data = pd.DataFrame({
    'Quarter': ['Q1', 'Q2', 'Q3', 'Q4'],
    'Value': [120, 150, 135, 180]
})
data['Cumulative value'] = data['Value'].cumsum()

display(data)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(data['Quarter'], data['Cumulative value'], marker='o')
ax.set_title('Second update: cumulative quarterly values')
ax.set_xlabel('Quarter')
ax.set_ylabel('Cumulative value')
ax.grid(True, alpha=0.3)

OUTDIR = '/content/demo_outputs'
os.makedirs(OUTDIR, exist_ok=True)
chart_path = os.path.join(OUTDIR, 'demo_cumulative_values_chart.png')
fig.savefig(chart_path, dpi=150, bbox_inches='tight')
plt.show()

print('Second update cell ran successfully.')
print('Chart saved inside this Colab run at:', chart_path)
print('Run time, UTC:', datetime.now(timezone.utc).strftime('%Y-%m-%d %H:%M:%S'))
```
